```python
import pandas as pd
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
# (sample, dimension) = (1797, 64)
digits.data
# Sample = Answer
digits.target
# 8x8 image
print(digits.feature_names)
# Digit names (0~9)
print(digits.target_names)
```

```python
df = pd.DataFrame(digits.data, columns=digits.feature_names)
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn import set_config
set_config(display="text")
def support_vector_machine(df: pd.DataFrame, target: pd.Series):
    feature_names = df.columns
    X_train, X_test, y_train, y_test = train_test_split(df, target, test_size=0.2, random_state=42)
    model = SVC(kernel='linear', C=1.0, random_state=42)
    model.fit(X_train, y_train)
    accuracy = model.score(X_test, y_test)
    print(f"Support Vector Machine Accuracy: {accuracy:.4f}")
    confusion_matrix = pd.crosstab(y_test, model.predict(X_test), rownames=['Actual'], colnames=['Predicted'])
    print("Confusion Matrix:\n", confusion_matrix)
    return model
support_vector_machine(df, digits.target)
```
