```python
import pandas as pd
from matplotlib import pyplot as plt
%matplotlib inline
from sklearn.model_selection import train_test_split
import seaborn as sns
```

```python
df = pd.read_csv('HR_comma_sep.csv')
df.head()
```

```python
df_left = df.groupby('left')
df_left.head()
```

```python
encoded_df = pd.get_dummies(df, columns=['Department', 'salary'], dtype=int)
encoded_df.head()
```

```python
correlation = encoded_df.corr()
correlation['left'].sort_values(key=lambda x: abs(x), ascending=False)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
def logistic_classification(df: pd.DataFrame):
    model = LogisticRegression()
    X = df[['satisfaction_level', 'Work_accident', 'time_spend_company', 'salary_low', 'salary_high']]
    y = df['left']
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.8, random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    print(f"Model coefficient: {model.coef_}")
    print(f"Model score: {accuracy_score(y_pred, y)}")
    print(f"Model Confusion matrix: {confusion_matrix(y_pred=y_pred, y_true=y)}")
```
