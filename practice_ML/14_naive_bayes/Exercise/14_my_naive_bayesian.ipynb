```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
def count_words(df: pd.DataFrame, y) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """Calculates the occurence of each word to conduct Multinominal Naive Bayesian
    Args:
        df: wine label data for categorizing wine types
    Returns:
        extracted words
    """
    vectorizer = CountVectorizer()
    X_train, y_train, X_test, y_test = train_test_split(df, y, test_size=0.2)
    X_train_vector = vectorizer.fit_transform(X_train, y)
    X_test_vector = vectorizer.transform(X_test)
    return (X_train_vector, y_train, X_test_vector, y_test)

import pandas as pd  # pandas
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.naive_bayes import MultinomialNB
import numpy as np  # NumPy
    
def naive_bayesian(df: pd.DataFrame, y):
    """Conducts Multinomial Naive Bayes """
    # Correct the unpacking order of the train-test split
    X_train, X_test, y_train, y_test = train_test_split(df, y)
        
    model = MultinomialNB()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
        
    # Calculate accuracy
    accuracy = np.mean(y_pred == y_test)
    print(f"Accuracy of the prediction: {accuracy}")
    # Run cross-validation
    scores = cross_val_score(model, X_train, y_train, cv=3)  # cv (Cross-Validation)
    print("Cross-Validation scores for each fold:", scores)
```

```python
naive_bayesian(df, y)
```

```python
from sklearn.pipeline import Pipeline
def naive_bayesian_pipeline(df: pd.DataFrame, y):
    """Makes a pipeline of calculating Bag of words to Naive Bayesian prediction"""
    bayesian_pipeline = Pipeline(steps=[
        ('vectorizer', CountVectorizer()),
        ('naive bayes', MultinomialNB())
    ])
    X_train, y_train, X_test, y_test = train_test_split(df, y, test_size=0.2)
    bayesian_pipeline.fit(X_train, y_train)
    y_pred = bayesian_pipeline.predict(X_test)
    accuracy = np.mean(np.array(y_pred == y_test))
    print(f"Accuracy of the prediction: {accuracy}")
    scores = cross_val_score(bayesian_pipeline, X_train, y_train, cv=3)
    print("Cross Validation scores for each fold", scores)
```
