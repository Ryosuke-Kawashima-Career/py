```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
def count_words(df: pd.DataFrame, y) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """Calculates the occurence of each word to conduct Multinominal Naive Bayesian
    Args:
        df: wine label data for categorizing wine types
    Returns:
        extracted words
    """
    vectorizer = CountVectorizer()
    X_train, y_train, X_test, y_test = train_test_split(df)
    X_train_vector = vectorizer.fit_transform(X_train, y, test_size=0.2)
    X_test_vector = vectorizer.transform(X_test)
    return (X_train_vector, y_train, X_test_vector, y_test)

from sklearn.model_selection import cross_val_score
import numpy as np
def naive_bayesian(df: pd.DataFrame, y):
    """Conducts Multinomincal Naive Bayesian
    Note. Naive Bayesian is used for spam filtering and sentimental analysis by calculating the distribution of word occurences
    """
    X_train, y_train, X_test, y_test = count_words(df, y)
    model = MultinomialNB()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = np.mean(np.array(y_pred == y_test))
    print(f"Accuracy of the prediction: {accuracy}")
    scores = cross_val_score(model, X_train, y_train, cv=3)
    print("Cross Validation scores for each fold", scores)
```
